```python
import numpy as np
import pandas as pd
import sklearn.model_selection
```

```python
data = pd.read_excel('../data/processed/cleaned_disaster_dataset.xlsx')
data
```

```python
data_train, data_test = sklearn.model_selection.train_test_split(
    data,
    train_size   = 0.70,
    shuffle      = True ,
    stratify=data["disaster_type"],
    random_state = 0
)
```

```python
data_train.to_excel('../data/processed/disaster_prediction_train.xlsx', index=False)
data_test.to_excel('../data/processed/disaster_prediction_test.xlsx', index=False)
```
